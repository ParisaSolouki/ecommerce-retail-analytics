# Project Definition

## Project Background

Olist is a Brazilian e-commerce marketplace that connects customers with sellers across different product categories and geographic regions.

The available dataset contains information about customers, orders, products, sellers, payments, reviews, and delivery timelines.

This project uses the available transactional data to evaluate business performance, customer experience, sales patterns, and operational efficiency.

## Business Goal

The primary goal of this project is to analyze Olist's e-commerce operations and generate actionable business insights that support data-driven decision making.

The analysis focuses on understanding sales performance, customer behavior, seller performance, payment patterns, delivery efficiency, and customer satisfaction.

The final deliverable is an interactive Power BI dashboard that enables business stakeholders to monitor key performance indicators and identify opportunities for improvement.

## Business Questions

The analysis addresses the following business questions:

**Q1. How did sales performance change over time?**

Monthly revenue, order volume, average order value, and year-over-year growth are analyzed to determine the main drivers of sales performance.

---

**Q2. Which product categories generated the highest sales volume and revenue?**

Product categories are compared by revenue, items sold, order volume, and average item price to identify the strongest-performing categories.

---

**Q3. Which sellers contributed most to the marketplace, and how did their performance vary?**

Seller revenue, order volume, active-seller count, and revenue concentration are evaluated to understand seller contribution and marketplace performance.

---

**Q4. How efficiently were orders fulfilled and delivered?**

Fulfillment time, shipping time, total delivery time, on-time delivery rate, and delay duration are analyzed to identify operational strengths and bottlenecks.

---

**Q5. How satisfied were customers, and how did delivery performance relate to review scores?**

Review scores, positive and negative review rates, delivery status, and product-category performance are analyzed to understand customer satisfaction.

---

**Q6. Which payment methods were most commonly used by customers?**

Payment-method usage, order share, payment value, and installment behavior are analyzed to understand customer payment preferences.

## Data Requirements

Based on the defined business questions and KPIs, only the data required for the planned analyses will be prepared and cleaned.

This project follows a business-driven data preparation approach. Instead of cleaning the entire dataset, only the tables and columns necessary to answer the business questions will be processed.
