# Project Definition

## Project Background

Olist is a Brazilian e-commerce marketplace that connects customers with sellers across different product categories and geographic regions.

The available dataset contains information about customers, orders, products, sellers, payments, reviews, and delivery timelines.

This project uses the available transactional data to evaluate business performance, customer experience, sales patterns, and operational efficiency.

## Business Goal

The primary goal of this project is to analyze Olist's e-commerce operations and generate actionable business insights that support data-driven decision making.

The analysis focuses on understanding sales performance, customer behavior, seller performance, payment patterns, delivery efficiency, and customer satisfaction.

The final outcome of the project will be an interactive Power BI dashboard that enables business stakeholders to monitor key performance indicators and identify opportunities for improvement.

## Business Questions

The analysis aims to answer the following business questions:

**Q1. How has sales performance changed over time?**

Understanding monthly sales trends helps identify growth patterns, seasonality, and periods of unusually high or low demand.

---

**Q2. Which product categories generate the highest sales and revenue?**

Identifying high-performing categories helps the business prioritize inventory, marketing efforts, and strategic investments.

---

**Q3. Which sellers contribute the most to the marketplace, and how does their performance vary?**

Evaluating seller performance supports decisions related to seller engagement, quality improvement, and marketplace management.

---

**Q4. How efficient is the order fulfillment and delivery process?**

Analyzing delivery times helps identify operational bottlenecks and opportunities to improve customer experience.

---

**Q5. How satisfied are customers with their purchases?**

Review scores provide valuable insight into customer satisfaction and overall service quality.

---

**Q6. Which payment methods are most commonly used by customers?**

Understanding payment preferences helps support financial planning and improve the checkout experience.

## Data Requirements

Based on the defined business questions and KPIs, only the data required for the planned analyses will be prepared and cleaned.

This project follows a business-driven data preparation approach. Instead of cleaning the entire dataset, only the tables and columns necessary to answer the business questions will be processed.
